```python
produtos = ["Camiseta Básica", "Calça Jeans", "Jaqueta Corta-Vento"]
precos = [49.90, 129.90, 219.90]

print("**** Cadastre um novo produto ****\n")
novo_produto = input("Digite o nome do produto: ")
novo_preco = float(input("Digite o preço do produto: \n"))

produtos.append(novo_produto)
precos.append(novo_preco)

print("Produto cadastrado com sucesso!\n")

for i,j in zip(produtos, precos):
  print(f"{i} custa R$ {j}")

if len(produtos) > 0:
  valor = int(input("\nDeseja excluir um produto cadatrado?: 1 = Sim 0 = Não\n"))
  if valor == 1:
    produto_removido = produtos.pop()
    preco_removido = precos.pop()
    print(f"O {produto_removido} de R$ {preco_removido} foi removido com sucesso!\n")
  else:
    print("Operação cancelada.")

else:
  print("Não há produtos cadastrados.")

for i,j in zip(produtos, precos):
  print(f"{i} custa R$ {j}")



```
